```python
import pandas as pd
import os
from difflib import get_close_matches

output_folder = r"C:\Users\LENOVO\cleaned_data"

file_path = os.path.join(
    output_folder,
    "Employees_cleaned.csv"
)

df = pd.read_csv(file_path)

# Correct department names already present in the dataset
standard_departments = [
    "Research & Development",
    "Finance & Accounting",
    "Information Technology",
    "Executive Leadership",
    "Human Resources",
    "Cybersecurity",
    "Legal & Compliance",
    "Operations",
    "Marketing",
    "Sales",
    "Customer Support",
    "Quality Assurance",
    "Facilities"
]

def correct_department(value):
    if pd.isna(value) or value == "Unknown":
        return value

    value = str(value).strip()

    # Exact match
    if value in standard_departments:
        return value

    # Find closest standard department
    match = get_close_matches(
        value,
        standard_departments,
        n=1,
        cutoff=0.70
    )

    if match:
        return match[0]

    return value

df["DepartmentName"] = df["DepartmentName"].apply(
    correct_department
)

df.to_csv(file_path, index=False)

print("DepartmentName standardization completed.")
print("Unique DepartmentName values after cleaning:",
      df["DepartmentName"].nunique())

print("\nFinal Department Names:")
print(df["DepartmentName"].dropna().unique())
```

```python
import pandas as pd
import os
from difflib import get_close_matches

output_folder = r"C:\Users\LENOVO\cleaned_data"

file_path = os.path.join(
    output_folder,
    "Employees_cleaned.csv"
)

df = pd.read_csv(file_path)

# Complete list of valid role names
standard_roles = [
    "Software Engineer",
    "Senior Software Engineer",
    "Financial Analyst",
    "Systems Administrator",
    "Chief Executive Officer",
    "Chief Financial Officer",
    "HR Manager",
    "HR Generalist",
    "Recruiter",
    "Accountant",
    "Account Manager",
    "SOC Engineer",
    "Compliance Officer",
    "Security Analyst",
    "Finance Manager",
    "Sales Director",
    "Sales Executive",
    "Content Strategist",
    "IT Support Technician",
    "Legal Counsel",
    "Marketing Specialist",
    "Network Engineer",
    "Office Administrator",
    "Operations Manager",
    "Operations Analyst",
    "Procurement Officer",
    "QA Engineer",
    "Support Agent",
    "Support Team Lead",
    "Facilities Coordinator",
    "Unknown"
]

def correct_role(role):
    if pd.isna(role):
        return "Unknown"

    role = str(role).strip()

    # Keep already correct roles
    if role in standard_roles:
        return role

    # Find closest standard role
    match = get_close_matches(
        role,
        standard_roles,
        n=1,
        cutoff=0.75
    )

    if match:
        return match[0]

    # Keep unmatched value for manual checking
    return role


df["RoleTitle"] = df["RoleTitle"].apply(correct_role)

df.to_csv(file_path, index=False)

print("RoleTitle cleaning completed.")

print(
    "\nNumber of unique RoleTitle values:",
    df["RoleTitle"].nunique()
)

print("\nFinal RoleTitle values:")
print(
    sorted(df["RoleTitle"].dropna().unique())
)
```

```python
import pandas as pd
import os
from difflib import get_close_matches

output_folder = r"C:\Users\LENOVO\cleaned_data"

file_path = os.path.join(
    output_folder,
    "PhishingCampaigns_cleaned.csv"
)

campaigns = pd.read_csv(file_path)

standard_types = [
    "Scheduled",
    "Random",
    "Targeted-High-Risk",
    "Unknown"
]

def clean_campaign_type(value):

    if pd.isna(value):
        return "Unknown"

    value = str(value).strip()

    # Exact match ignoring case
    for standard in standard_types:
        if value.lower() == standard.lower():
            return standard

    # Find closest standard type
    match = get_close_matches(
        value,
        standard_types,
        n=1,
        cutoff=0.65
    )

    if match:
        return match[0]

    return "Unknown"


campaigns["CampaignType"] = campaigns["CampaignType"].apply(
    clean_campaign_type
)

campaigns.to_csv(file_path, index=False)

print("CampaignType cleaning completed.")

print("\nCampaignType values after cleaning:")
print(
    campaigns["CampaignType"]
    .value_counts()
)
```

```python
import pandas as pd

file = r"C:\Users\LENOVO\Phishing_Simulation_Data_CLEANED.xlsx"

simulation = pd.read_excel(file, sheet_name="SimulationResults")

print("Total records:", len(simulation))

print("\nRisk Score Summary:")
print(simulation["RiskScore"].describe())

print("\nRisk Score Sample:")
print(simulation["RiskScore"].head(10))
```

```python
# Create Risk Category based on RiskScore

def create_risk_category(row):
    risk_score = row["RiskScore"]

    if pd.isna(risk_score):
        return "Unknown"

    if risk_score >= 30:
        return "High Risk"
    elif risk_score >= 10:
        return "Medium Risk"
    else:
        return "Low Risk"


simulation["RiskCategory"] = simulation.apply(
    create_risk_category,
    axis=1
)

print(simulation["RiskCategory"].value_counts())
```

```python
employees = pd.read_excel(file, sheet_name="Employees")

# Standardize EmployeeID
simulation["EmployeeID"] = simulation["EmployeeID"].astype(str).str.upper().str.strip()
employees["EmployeeID"] = employees["EmployeeID"].astype(str).str.upper().str.strip()

# Keep employee details
employee_details = employees[
    ["EmployeeID", "DepartmentName", "RoleTitle"]
].drop_duplicates("EmployeeID")

# Merge employee details with risk analysis
risk_analysis = simulation.merge(
    employee_details,
    on="EmployeeID",
    how="left"
)

# High-risk records only
high_risk = risk_analysis[
    risk_analysis["RiskCategory"] == "High Risk"
].copy()

print("High Risk Records:", len(high_risk))

print("\nHigh Risk by Department:")
print(high_risk["DepartmentName"].value_counts())

print("\nHigh Risk by Role:")
print(high_risk["RoleTitle"].value_counts().head(15))
```

```python
# High Risk percentage by Department

department_risk = risk_analysis.groupby("DepartmentName").agg(
    Total_Records=("ResultID", "count"),
    High_Risk=("RiskCategory", lambda x: (x == "High Risk").sum())
).reset_index()

department_risk["High_Risk_%"] = (
    department_risk["High_Risk"] /
    department_risk["Total_Records"] * 100
)

department_risk = department_risk.sort_values(
    "High_Risk_%",
    ascending=False
).round(2)

print("High Risk Percentage by Department:")
print(department_risk)
```

```python
# Task 13 - Department-Wise Phishing Analysis

department_analysis = risk_analysis.groupby("DepartmentName").agg(
    Total_Simulations=("ResultID", "count"),
    Clicks=("LinkClickedFlag", "sum"),
    Credential_Submissions=("CredentialsEnteredFlag", "sum"),
    Reports=("ReportedAsPhishingFlag", "sum"),
    Average_Risk_Score=("RiskScore", "mean")
).reset_index()

# Calculate rates
department_analysis["Click_Rate_%"] = (
    department_analysis["Clicks"] /
    department_analysis["Total_Simulations"] * 100
)

department_analysis["Credential_Submission_Rate_%"] = (
    department_analysis["Credential_Submissions"] /
    department_analysis["Total_Simulations"] * 100
)

department_analysis["Report_Rate_%"] = (
    department_analysis["Reports"] /
    department_analysis["Total_Simulations"] * 100
)

# Round the results
department_analysis = department_analysis.round(2)

# Sort by Click Rate
department_analysis = department_analysis.sort_values(
    "Click_Rate_%",
    ascending=False
)

print(department_analysis.to_string(index=False))
```

```python
import pandas as pd
import os

# Your final 15-sheet workbook
input_file = r"C:\Users\LENOVO\Downloads\clean it data\Phishing_Simulation_Data_FINAL.xlsx"

# New output file
output_file = r"C:\Users\LENOVO\Downloads\clean it data\Phishing_Simulation_Data_FINAL_NO_BLANKS.xlsx"

# Read all sheets
data = pd.read_excel(input_file, sheet_name=None)

for sheet_name, df in data.items():

    print("\nProcessing:", sheet_name)

    for col in df.columns:

        # Numeric columns → Median
        if pd.api.types.is_numeric_dtype(df[col]):
            if df[col].isna().sum() > 0:
                median_value = df[col].median()
                df[col] = df[col].fillna(median_value)

        # Date / datetime columns → Not Recorded
        elif pd.api.types.is_datetime64_any_dtype(df[col]):
            df[col] = df[col].fillna("Not Recorded")

        # Text / categorical columns → Unknown
        else:
            df[col] = df[col].fillna("Unknown")

    data[sheet_name] = df

# Save new workbook
with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    for sheet_name, df in data.items():
        df.to_excel(
            writer,
            sheet_name=sheet_name[:31],
            index=False
        )

print("\n✅ ALL BLANKS HANDLED!")
print("File created:")
print(output_file)
```

```python
# TASK 7 – Validate and Standardize Phishing Flags

flag_columns = {
    "RoleAssignmentHistory": ["IsCurrentRole", "PromotionFlag"],
    "LocationAccessLog": ["AccessGranted", "VisitorFlag"],
    "PhishingTemplates": ["SenderSpoofed", "ApprovedFlag"],
    "SimulationResults": [
        "EmailOpenedFlag",
        "LinkClickedFlag",
        "CredentialsEnteredFlag",
        "AttachmentDownloadedFlag",
        "ReportedAsPhishingFlag",
        "TrainingAssignedFlag"
    ],
    "TrainingRecords": ["CertificateIssued"],
    "DeviceInventory": [
        "AntivirusInstalled",
        "FirewallEnabled",
        "EncryptionEnabled",
        "IsCompliant"
    ],
    "SecurityAwarenessScores": ["FollowUpRequired"],
    "HelpdeskTickets": ["SLABreached", "ReopenedFlag"],
    "VulnerabilityData": ["PatchedFlag"],
    "LoginActivityLog": ["MFAUsed", "VPNUsed", "GeoAnomalyFlag"],
    "EmailTrafficLog": ["ExternalSenderFlag", "EncryptedFlag"]
}

flag_map = {
    "yes": 1,
    "y": 1,
    "true": 1,
    "1": 1,

    "no": 0,
    "n": 0,
    "false": 0,
    "0": 0,

    "unknown": -1
}

for sheet_name, columns in flag_columns.items():

    df = data[sheet_name].copy()

    for col in columns:

        if col in df.columns:

            # Convert values to standard 0 / 1 / -1
            df[col] = (
                df[col]
                .astype("string")
                .str.strip()
                .str.lower()
                .map(flag_map)
            )

    data[sheet_name] = df

print("✅ TASK 7 FLAG STANDARDIZATION COMPLETED")
```
